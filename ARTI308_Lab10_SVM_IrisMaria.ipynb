# Dataset Description
The Iris dataset contains 150 flower samples from three species:
- Setosa   Versicolor  Virginica

Each sample has 4 numerical features:
- Sepal length
- Sepal width
- Petal length
- Petal width

The target variable is the flower species.


## 2. Import Required Libraries


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

```

## 3. Load the Dataset


```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
iris.info()
```

```python
iris.describe()
```

# 4 (EDA)



```python
sns.pairplot(iris, hue='species', palette='Set1')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue='species')
plt.title('Petal Length vs Petal Width')
plt.show()
```

```python
setosa = iris[iris['species'] == 'setosa']

plt.figure(figsize=(8, 6))
sns.kdeplot(x=setosa['sepal_width'], y=setosa['sepal_length'], fill=True, cmap='plasma')
plt.title('KDE Plot for Iris Setosa')
plt.show()
```

## 5. Data Preprocessing


```python
X = iris.drop('species', axis=1)
y = iris['species']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=101
)
```

## 6. Train the SVM Model


```python
model = SVC()
model.fit(X_train, y_train)
```

## 7. Make Predictions


```python
predictions = model.predict(X_test)
predictions[:10]
```

## 8. Evaluate the Model



```python
print('Accuracy:', accuracy_score(y_test, predictions))
print('\nConfusion Matrix:')
print(confusion_matrix(y_test, predictions))
print('\nClassification Report:')
print(classification_report(y_test, predictions))
```

## 9. Hyperparameter Optimization Using GridSearchCV

GridSearchCV is used to find the best values for SVM parameters such as `C`, `gamma`, and `kernel`.


```python
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf']
}

grid = GridSearchCV(SVC(), param_grid, refit=True, verbose=2)
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid_predictions = grid.predict(X_test)

print('GridSearchCV Accuracy:', accuracy_score(y_test, grid_predictions))
print('\nConfusion Matrix:')
print(confusion_matrix(y_test, grid_predictions))
print('\nClassification Report:')
print(classification_report(y_test, grid_predictions))
```

## 10. Conclusion

The SVM model showed high performance in classifying the Iris dataset. The visual analysis indicated that the different Iris species are clearly distinguishable, particularly based on petal length and petal width. By using GridSearchCV, the model’s parameters were tuned to enhance classification accuracy and minimize prediction mistakes.
